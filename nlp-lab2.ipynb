```python
import numpy as np 
import pandas as pd
from nltk.corpus import brown, stopwords
from nltk import word_tokenize, sent_tokenize
import nltk
import string
```

```python
brown_sentences = brown.sents()
sentences = brown_sentences[:20]
print("Number of sentences selected:", len(sentences))
```

```python
for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}:")
    print(" ".join(sentence))
    print()
```

```python
lowercase_sentences = []

for sentence in sentences:

    lowercase_tokens = [
        word.lower()
        for word in sentence
    ]

    lowercase_sentences.append(lowercase_tokens)

print(lowercase_sentences)
```

```python
stop_words = set(stopwords.words('english'))

cleaned_sentences = []

for sentence in lowercase_sentences:

    cleaned_tokens = [
        word
        for word in sentence
        if word not in string.punctuation
        and word not in stop_words
        and word.isalpha()
    ]

    cleaned_sentences.append(cleaned_tokens)

print(cleaned_sentences)
```
